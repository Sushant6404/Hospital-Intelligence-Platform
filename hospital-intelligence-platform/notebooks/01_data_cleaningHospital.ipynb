```python
duplicates = df[df.duplicated()]

duplicates.head()
```

```python
df = df.drop_duplicates().reset_index(drop=True)

print("New Shape:", df.shape)
```

```python
print("Duplicate Records:", df.duplicated().sum())
```

```python
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
)

df.columns
```

```python
df["date_of_admission"] = pd.to_datetime(df["date_of_admission"])

df["discharge_date"] = pd.to_datetime(df["discharge_date"])
```

```python
# Create Age Groups
bins = [0,18,35,50,65,100]

labels = [
    "Child",
    "Young Adult",
    "Adult",
    "Senior",
    "Elderly"
]

df["age_group"] = pd.cut(
    df["age"],
    bins=bins,
    labels=labels
)
df.head()
```

```python
df["age_group"].value_counts()
```

```python
df["admission_year"] = df["date_of_admission"].dt.year

df["admission_month"] = df["date_of_admission"].dt.month_name()

df["admission_day"] = df["date_of_admission"].dt.day_name()

df["admission_quarter"] = "Q" + df["date_of_admission"].dt.quarter.astype(str)

df.head()
```

```python
df["date_of_admission"] = pd.to_datetime(df["date_of_admission"])

df["discharge_date"] = pd.to_datetime(df["discharge_date"])
```

```python
df.info()
```

```python
df["length_of_stay"] = (
    df["discharge_date"] -
    df["date_of_admission"]
).dt.days
```

```python
df["length_of_stay"].describe()
```

```python
df["billing_amount"].describe()
```

```python
df[df["billing_amount"] < 0]
```

```python
negative_count = (df["billing_amount"] < 0).sum()

print("Negative Billing Records:", negative_count)
```

```python
billing_median = df.loc[
    df["billing_amount"] >= 0,
    "billing_amount"
].median()

print("Median Billing Amount:", round(billing_median, 2))
```

```python
df.loc[
    df["billing_amount"] < 0,
    "billing_amount"
] = billing_median
```

```python
print("Negative Billing Records:",
      (df["billing_amount"] < 0).sum())
```

```python
# Standardize Text Columns
text_columns = [
    "name",
    "doctor",
    "hospital"
]

for col in text_columns:
    df[col] = (
        df[col]
        .str.strip()
        .str.title()
    )
```
